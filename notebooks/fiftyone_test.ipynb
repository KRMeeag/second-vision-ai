# fiftyone_test.ipynb — original scratch exploration (historical)

**What this is:** the student's original, hand-written FiftyOne experiment — pulling a small Trash Bins ("Waste container") sample directly from the Open Images Zoo to see what the data actually looked like. This is what first surfaced `IsDepiction` as something worth filtering (see `docs/DECISIONS.md`).

**Status:** kept as-is for historical reference, not part of the maintained notebook tooling. For day-to-day work, use one of the three purpose-built notebooks instead:

| Notebook | When to use it |
|---|---|
| `fiftyone_preview.ipynb` | **Before pulling** — preview a class at real volume straight from the Zoo, to tune filters before running the real acquire script. Network required, nothing saved to disk. |
| `fiftyone_explore.ipynb` | **After a raw pull** — browse an already-acquired raw per-class export (`dataset/raw/open_images/<class>/`, COCO-style). No network needed. |
| `fiftyone_review_processed.ipynb` | **After conversion** — browse converted intermediate-schema output (`dataset/processed/<source>/`, canonical class ids, flat `images/`+`labels/`). This is what Stage 5.3/5.5 curation actually looks at. |

No filesystem paths depend on this notebook's location, so it didn't need any path fixes when moved into `notebooks/`.

```python
import fiftyone as fo
import fiftyone.zoo as foz
from fiftyone import ViewField as F
```

```python
print(fo.list_datasets())
```

```python
dataset = foz.load_zoo_dataset(
    "open-images-v7",
    split="validation",
    dataset_name="sv-common-v1",
    classes=["Waste container"],
    label_types=["detections"],
    max_samples=200,
    shuffle=True,
    seed=42,
    drop_existing_dataset=True,
)
```

```python
view = dataset.filter_labels(
    "ground_truth",
    F("IsDepiction") == False,
)
```

```python
session = fo.launch_app(view, auto=False)
```
