```python
KEY  = "bench"      # "bench" or "stairs"
MODE = "clean"      # "clean" | "groupof" | "all"

import fiftyone as fo
from fiftyone import ViewField as F

LABEL = {"bench": "Bench", "stairs": "Stairs"}[KEY]
ds = fo.load_dataset(f"preview_oi_{KEY}")

if MODE == "all":
    view = ds
else:
    cond = (F("label") == LABEL) & (F("IsDepiction") == False)
    cond = cond & (F("IsGroupOf") == (MODE == "groupof"))
    view = ds.filter_labels("ground_truth", cond) \
             .match(F("ground_truth.detections").length() > 0)

print(f"{KEY}/{MODE}  label={LABEL}  "
      f"{len(view)} images, {view.count('ground_truth.detections')} boxes")
session = fo.launch_app(view, auto=False)
```
